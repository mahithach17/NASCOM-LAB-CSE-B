```python
import random
from datetime import datetime, timedelta

DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Light_Sensor']
LOCATIONS = ['Lab1', 'Lab2', 'OfficeA', 'OfficeB', 'Warehouse']
READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),   # Celsius
    'Humidity_Sensor': (30.0, 80.0),      # Percentage
    'Pressure_Sensor': (980.0, 1030.0),   # hPa
    'Light_Sensor': (100, 1000)           # Lux
}

print("\n--- Task 1: Data Generation and Basic Structures ---")

raw_sensor_data = []

# Starting time
start_time = datetime(2026, 7, 20, 10, 0, 0)

for i in range(100):
    timestamp = start_time + timedelta(minutes=i)
    timestamp_str = timestamp.strftime("%Y-%m-%d %H:%M:%S")

    device_id = f"DEV{i+1:03d}"      # DEV001, DEV002, ...

    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    low, high = READING_RANGES[device_type]
    value = round(random.uniform(low, high), 2)

    reading = (timestamp_str, device_id, device_type, location, value)
    raw_sensor_data.append(reading)

# Display first 5 readings
for reading in raw_sensor_data[:100]:
    print(reading)
```

```python
def get_readings_by_location(readings_list, target_location):
    result = []

    for reading in readings_list:
        if reading[3] == target_location:
            result.append(reading)

    return result
lab1_readings = get_readings_by_location(raw_sensor_data, "Lab1")

for i in lab1_readings:
    print(i)
```

```python
import os
import datetime
import random

print("\n--- Task 3: Core Library Usage ---")

reports_dir = "sensor_reports"

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"Directory '{reports_dir}' created.")
else:
    print(f"Directory '{reports_dir}' already exists.")


if len(raw_sensor_data) > 1:

    first_timestamp = raw_sensor_data[0][0]
    last_timestamp = raw_sensor_data[-1][0]

    date_format = "%Y-%m-%d %H:%M:%S"

    first_datetime = datetime.datetime.strptime(first_timestamp, date_format)
    last_datetime = datetime.datetime.strptime(last_timestamp, date_format)

    duration = last_datetime - first_datetime

    days = duration.days
    total_seconds = duration.total_seconds()
    hours = int(total_seconds // 3600) % 24
    minutes = int((total_seconds % 3600) // 60)

    print(f"Duration: {days} days, {hours} hours, {minutes} minutes.")

else:
    print("Not enough data to calculate duration.")

random_values = []

for i in range(5):
    random_values.append(random.randint(100, 200))

print("Random Values:", random_values)
```
