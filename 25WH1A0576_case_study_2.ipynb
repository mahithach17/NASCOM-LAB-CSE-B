```python
import pandas as pd
import numpy as np
data = {
    'Student': ['S1','S2','S3','S4','S5','S6','S7','S8','S9','S10'],
    'Study_Hours': [2,3,4,5,6,7,8,4,6,9],
    'Attendance': [65,70,75,80,85,90,92,78,88,95],
    'Assignment _score':[55,60,65,72,78,85,88,70,80,92],
    'Previous_Score':[58,62,68,70,76,82,86,72,79,80],
    'Final_Score':[60,64,70,75,82,88,91,74,84,94]
}
df = pd.DataFrame(data)
```

```python
#----------------Task-2--------------------


print("\nDataFrame:")
print(df)
print("\n\n")
print("Average Study Hours :")
print(df['Study_Hours'].mean())
print("\n\n")
print("Median Attendance:")
print(df['Attendance'].median())
print("\n\n")
print("Average Assignment Score")
print(df['Assignment _score'].mean())
print("\n\n")
print("Average Previous Score")
print(df['Previous_Score'].mean())
print()
print("Average Final Score:")
print(df['Final_Score'].max())
print("\n\n")
high_final=df[df['Final_Score']==df['Final_Score'].max()]
print("Student with highest final score:")
print(high_final)
print("\n\n")
low_final=df[df['Final_Score']==df['Final_Score'].min()]
print("Student with lowest final score:")
print(low_final)
print("\n\n")
attendance_80=df[df['Attendance']>80]
print("Students with attendance greater than 80:")
print(attendance_80)
print("\n\n")
final_score_80=df[df['Final_Score']>80]
print("Students with final score greater than 80:")
print(final_score_80)
print("\n\n")
summary=df.describe()
print(summary)
```

```python
#-------------TASK-1---------------
num_students = len(df['Student'])
print(f"Number of students: {num_students}")
print("\n\n")
print("Column names:")
print(df.columns)
print("\n\n")
print("\nFirst five rows:")
print(df.iloc[0:5])
print("\n\n")
print("\nlast five rows:")
print(df.tail(5))
```

```python
#--------------TASK-3----------------
final_scores = df['Final_Score'].values
mean_final_score = np.mean(final_scores)
median_final_score = np.median(final_scores)
max_final_score = np.max(final_scores)
min_final_score = np.min(final_scores)
std_dev_final_score = np.std(final_scores)
print(f"Final Scores (NumPy Array): {final_scores}")
print(f"Mean Final Score: {mean_final_score:.2f}")
print(f"Median Final Score: {median_final_score:.2f}")
print(f"Maximum Final Score: {max_final_score:.2f}")
print(f"Minimum Final Score: {min_final_score:.2f}")
print(f"Standard Deviation of Final Scores: {std_dev_final_score:.2f}")
```

```python
#--------------task-4-----------------
import matplotlib.pyplot as plt
import seaborn as sns
# 1
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Study_Hours', y='Final_Score', data=df)
plt.title('Study Hours vs Final Score')
plt.xlabel('Study Hours')
plt.ylabel('Final Score')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
print("\n\n")
# 2
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Attendance', y='Final_Score', data=df)
plt.title('Attendance vs Final Score')
plt.xlabel('Attendance')
plt.ylabel('Final Score')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
print("\n\n")
# 3
plt.figure(figsize=(8, 6))
sns.histplot(df['Final_Score'], bins=5, kde=True)
plt.title('Distribution of Final Scores')
plt.xlabel('Final Score')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
print("\n\n")
# 4
plt.figure(figsize=(10, 6))
sns.barplot(x='Student', y='Final_Score', data=df.sort_values('Final_Score', ascending=False), palette='viridis')
plt.title('Student-wise Final Scores')
plt.xlabel('Student')
plt.ylabel('Final Score')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#---------------TASK-6-------------------
def F(x):
    return 40 + 5*x - 0.2*x**2
def dF(x):
    return 5 - 0.4*x
print("x\tF(x)\tdF(x)")
print("-----------------------")
for x in range(1, 11):
    print(f"{x}\t{F(x):.2f}\t{dF(x):.2f}")
x_critical = 5 / 0.4
print(f"\nValue of x for which F'(x) = 0: {x_critical:.2f}")
```

```python
#---------------------TASK-7---------------------
import numpy as np
def L(x):
    return 5*x - 0.2*x**2
a = 0
b = 5
n = 1000
x = np.linspace(a, b, n + 1)
y = L(x)
area = np.trapezoid(y, x)
print("Numerical approximation =", area)
exact = 2.5*(5**2) - (0.2*(5**3))/3
print("Exact value =", exact)
```
