```python
def f(x):
  return x**2
h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print(round(slope_at_3,4))
```

```python
import sympy as sp
x = sp.Symbol('x')
exp = x**2
deriv=sp.diff(exp,x)
print(deriv)
print(deriv.subs(x,3))
```
