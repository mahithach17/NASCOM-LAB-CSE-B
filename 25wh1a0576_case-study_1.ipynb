```python

#--------------------TASK-1---------------------#
def battery_status(Charge):
  if Charge >70:
    status="High"
  elif Charge <=70 and Charge>50:
      status="Medium"
  else:
      status="Low"
  return status
battery_data=[
    {'Time':0,'Temperature':32,'Voltage':400,'Current':20,'Charge':95},
    {'Time':1,'Temperature':35,'Voltage':395,'Current':22,'Charge':88},
    {'Time':2,'Temperature':38,'Voltage':390,'Current':25,'Charge':80},
    {'Time':3,'Temperature':40,'Voltage':385,'Current':28,'Charge':70},
    {'Time':4,'Temperature':42,'Voltage':380,'Current':30,'Charge':60},
    {'Time':5,'Temperature':45,'Voltage':375,'Current':32,'Charge':48}
]
for i in battery_data:
  print(f"Time: {i["Time"]} , Temperature: {i["Temperature"]} , Voltage: {i["Voltage"]} , Current: {i["Current"]} , Charge: {i["Charge"]}")
  if i["Temperature"]>40 :
    print("High Temperature Alert!")
  else:
    print("Temperature Normal")
  if i["Charge"]<50 :
    print("Low Battery Alert!")
  else:
    print("Battery Charge Normal")
  status = battery_status(i["Charge"])
  print(f"Time: {i['Time']}, Charge: {i['Charge']}, Status: {status}")
  print("\n")

```

```python
#--------------TASK-3------------------
import numpy as np
A = np.array([[2, 1], [1, 3]])
X = np.array([[40], [25]])
print("Matrix A:")
print(A)
print("\nVector X:")
print(X)
Y = np.dot(A, X)
print("\nResulting vector Y (Battery Stress Index):")
print(Y)
print("\n--- Additional Linear Algebra Tasks for Matrix A ---")
A_transpose = A.T
print("\nTranspose of A (A^T):")
print(A_transpose)
det_A = np.linalg.det(A)
print("\nDeterminant of A (det(A)):")
print(det_A)
if det_A != 0:
    A_inverse = np.linalg.inv(A)
    print("\nInverse of A (A^-1):")
    print(A_inverse)
else:
    print("\nInverse of A does not exist because the determinant is zero.")
print("\n--- Solving a System of Equations ---")
A_system = np.array([[2, 1], [1, 3]])
B_system = np.array([100, 120])
print("System Matrix A:")
print(A_system)
print("\nSystem Vector B:")
print(B_system)
X_solution = np.linalg.solve(A_system, B_system)
print("\nSolution for X (x, y):")
print(X_solution)
print(f"x = {X_solution[0]:.2f}")
print(f"y = {X_solution[1]:.2f}")
```

```python
#-----------------------TASK-4---------------------
import sympy as sp
t = sp.symbols('t')
C_t = 100 - 2*t - (1/2)*t**2
print(f"Battery charge function C(t): {C_t}")
dC_dt = sp.diff(C_t, t)
print(f"Rate of change of battery charge dC/dt: {dC_dt}")
discharge_rate_t2 = dC_dt.subs(t, 2)
print(f"Discharge rate at t = 2 hours (dC/dt at t=2): {discharge_rate_t2}")
discharge_rate_t5 = dC_dt.subs(t, 5)
print(f"Discharge rate at t = 5 hours (dC/dt at t=5): {discharge_rate_t5}")
if discharge_rate_t2 < discharge_rate_t5:
    interpretation = "The battery is losing charge at an increasing rate (the rate of change is becoming more negative)."
elif discharge_rate_t2 > discharge_rate_t5:
    interpretation = "The battery is losing charge at a decreasing rate (the rate of change is becoming less negative)."
else:
    interpretation = "The battery is losing charge at a constant rate."

print(interpretation)
```
